# 5 .merging data

```python
first = order_items.merge(products,on='product_id',how='outer')
```

```python
second = orders.merge(customers,on='customer_id',how='outer')
```

```python
df = first.merge(second,on='order_id',how='outer')
```

```python
# This is my final dataset where all the tables have been included using the merge() method in Pandas.
# The merge() function helps establish relationships between two or more tables based on common information among them,
# often in a time-based manner.
```

```python
df
```
